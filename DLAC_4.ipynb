```python

import os
import platform
import subprocess

print("Operating System:", platform.platform())
print("Python Version:", platform.python_version())
print("Working Directory:", os.getcwd())

git_version = subprocess.run(
    ["git", "--version"],
    capture_output=True,
    text=True
)

print("Git:", git_version.stdout.strip())
```

```python

from pathlib import Path

project_root = Path("/content/VisionPort")

directories = [
    "data",
    "notebooks",
    "src",
    "models",
    "results",
    "docs"
]

for directory in directories:
    (project_root / directory).mkdir(
        parents=True,
        exist_ok=True
    )

print("Project structure created.")

for path in sorted(project_root.iterdir()):
    print(path.name, "/", "Directory" if path.is_dir() else "File")
```

```python

os.chdir(project_root)

result = subprocess.run(
    ["git", "init"],
    capture_output=True,
    text=True
)

print(result.stdout)
print(result.stderr)
```

```python

git_name = input("Enter your Git name: ")
git_email = input("Enter your Git email: ")

subprocess.run(
    ["git", "config", "user.name", git_name],
    check=True
)

subprocess.run(
    ["git", "config", "user.email", git_email],
    check=True
)

print("Git identity configured.")
```

```python

readme_content = """
# VisionPort

Deep Learning inference and accelerated computer vision
development project focused on Indian traffic imagery.

## Project Structure

- data: datasets and image samples
- notebooks: experimental notebooks
- src: source code
- models: model artifacts
- results: inference outputs and measurements
- docs: technical documentation

## Development Environment

- Python
- PyTorch
- CUDA
- Google Colab
"""

(project_root / "README.md").write_text(
    readme_content.strip(),
    encoding="utf-8"
)

print("README created.")
print((project_root / "README.md").read_text())
```

```python

gitignore_content = """
# Python
__pycache__/
*.pyc
.ipynb_checkpoints/

# Virtual environments
venv/
.env
.venv/

# Model weights
*.pt
*.pth
*.onnx
*.engine

# Dataset
data/

# Results
results/

# Secrets
secrets.json
credentials.json
"""

(project_root / ".gitignore").write_text(
    gitignore_content.strip(),
    encoding="utf-8"
)

print(".gitignore created.")
```

```python

status = subprocess.run(
    ["git", "status"],
    capture_output=True,
    text=True
)

print(status.stdout)
```

```python

subprocess.run(
    ["git", "add", "README.md", ".gitignore"],
    check=True
)

print("Files staged successfully.")
```

```python

subprocess.run(
    ["git", "commit", "-m", "Initial VisionPort project structure"],
    check=True
)

print("Initial commit created.")
```

```python

history = subprocess.run(
    ["git", "log", "--oneline"],
    capture_output=True,
    text=True
)

print(history.stdout)
```

```python

import torch

print("PyTorch Environment")
print("=" * 40)

print("PyTorch Version:", torch.__version__)
print("CUDA Build Version:", torch.version.cuda)
print("CUDA Available:", torch.cuda.is_available())
```

```python

if torch.cuda.is_available():

    gpu = torch.cuda.get_device_properties(0)

    print("GPU Device:", torch.cuda.get_device_name(0))
    print("GPU Capability:", torch.cuda.get_device_capability(0))
    print("Total GPU Memory:", round(gpu.total_memory / (1024**3), 2), "GB")
    print("Multiprocessors:", gpu.multi_processor_count)

else:
    print("No CUDA-enabled GPU detected.")
```

```python

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

a = torch.randn(1000, 1000, device=device)
b = torch.randn(1000, 1000, device=device)

c = torch.matmul(a, b)

print("Tensor A Device:", a.device)
print("Tensor B Device:", b.device)
print("Output Device:", c.device)
print("Output Shape:", c.shape)
print("Execution Successful:", torch.isfinite(c).all().item())
```

```python

import sys
import json

environment_report = {
    "python_version": sys.version,
    "platform": platform.platform(),
    "pytorch_version": torch.__version__,
    "cuda_build_version": torch.version.cuda,
    "cuda_available": torch.cuda.is_available(),
    "gpu_name": (
        torch.cuda.get_device_name(0)
        if torch.cuda.is_available()
        else "No GPU"
    )
}

print(json.dumps(environment_report, indent=4))
```

```python

report_path = project_root / "docs" / "environment.json"

report_path.write_text(
    json.dumps(environment_report, indent=4),
    encoding="utf-8"
)

print("Environment report saved:", report_path)
```

```python

subprocess.run(
    ["git", "add", "docs/environment.json"],
    check=True
)

subprocess.run(
    ["git", "commit", "-m", "Record PyTorch CUDA environment"],
    check=True
)

print("Environment snapshot committed.")
```

```python

print("Repository Status")
print("=" * 40)

status = subprocess.run(
    ["git", "status"],
    capture_output=True,
    text=True
)

print(status.stdout)

print("\nCommit History")

history = subprocess.run(
    ["git", "log", "--oneline", "-5"],
    capture_output=True,
    text=True
)

print(history.stdout)

print("\nProject Files")

for path in sorted(project_root.rglob("*")):
    if ".git" not in path.parts:
        print(path.relative_to(project_root))
```
